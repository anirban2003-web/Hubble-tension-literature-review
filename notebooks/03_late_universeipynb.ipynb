```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("h0_late_universe_fig3_data.csv")

# Order exactly as in the reference figure
order = [
    "SH0ES Cepheids",
    "SBF (early-type)",
    "CCHP TRGB",
    "TDCOSMO lensing",
    "GW standard sirens"
]

df = df.set_index("probe").loc[order].reset_index()
```

```python
fig, ax = plt.subplots(
    figsize=(9, 3.2),
    dpi=100
)
# Top-to-bottom ordering
y = np.arange(len(df))[::-1]



xerr = np.vstack([
    df["error_minus"].values,
    df["error_plus"].values
])

ax.errorbar(
    df["H0"],
    y,
    xerr=xerr,

    fmt="o",

    # red circular points
    markersize=8,
    markerfacecolor="#d24a43",
    markeredgecolor="#a93631",
    markeredgewidth=0.7,

    # blue error bars
    ecolor="#5555ee",
    elinewidth=0.9,
    capsize=3.5,
    capthick=0.8,

    linestyle="none",
    zorder=4
)

PLANCK = 67.4

ax.axvline(
    PLANCK,
    color="#315a91",
    linestyle="--",
    linewidth=1.1,
    zorder=1
)

CCHP = 69.8

ax.axvline(
    CCHP,
    color="#999999",
    linestyle=":",
    linewidth=1.0,
    zorder=1
)

ax.text(
    63.0,
    -0.42,
    r"$Planck$",
    fontsize=12,
    color="#315a91",
    fontstyle="italic",
    ha="left",
    va="center"
)

ax.text(
    70.7,
    3.72,
    "CCHP intermediate",
    fontsize=10.5,
    color="#888888",
    ha="left",
    va="center"
)

labels = [
    r"SH0ES Cepheids $\mathbf{[3]}$",
    r"SBF (early-type) $\mathbf{[22]}$",
    r"CCHP TRGB $\mathbf{[21]}$",
    r"TDCOSMO lensing $\mathbf{[20]}$",
    r"GW standard sirens $\mathbf{[40]}$"
]

ax.set_yticks(y)
ax.set_yticklabels(
    labels,
    fontsize=11
)
ax.set_xlim(58, 85)

ax.set_xticks(
    np.arange(58, 86, 2)
)

ax.set_xlabel(
    r"$H_0\;(\mathrm{km\,s^{-1}\,Mpc^{-1}})$",
    fontsize=12,
    labelpad=4
)

ax.tick_params(
    axis="x",
    labelsize=9,
    direction="out",
    length=3
)

ax.tick_params(
    axis="y",
    length=0
)

ax.grid(
    axis="x",
    color="#d9d9d9",
    linewidth=0.55,
    alpha=0.75
)

ax.grid(
    axis="y",
    color="#e5e5e5",
    linewidth=0.5,
    alpha=0.65
)

ax.set_axisbelow(True)

ax.set_ylim(-0.5, 4.5)



ax.set_title(
    r"Late-Universe $H_0$ Measurements Contextualising SH0ES",
    fontsize=14,
    pad=10
)



ax.set_ylabel(
    "Late-Universe probe",
    fontsize=11,
    labelpad=11
)

for spine in ax.spines.values():
    spine.set_linewidth(0.75)
    spine.set_color("#555555")



ax.tick_params(
    axis="both",
    which="major",
    width=0.6
)


plt.subplots_adjust(
    left=0.445,
    right=0.965,
    top=0.82,
    bottom=0.25
)




plt.savefig(
    "figure3_late_universe_H0.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
