```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("mortsell_2022_fig2_data.csv")

# Values from the CSV
H0 = df["H0"].values
H0_err = df["error"].values
```

```python
fig, ax = plt.subplots(figsize=(9, 4.58), dpi=100)

# Background
ax.set_facecolor("white")
x = np.arange(len(df))

ax.errorbar(
    x,
    H0,
    yerr=H0_err,
    fmt="D",                 # diamond marker
    markersize=7,
    markerfacecolor="#3569a8",
    markeredgecolor="#1f4f82",
    markeredgewidth=0.8,

    # error bars
    ecolor="blue",
    elinewidth=0.8,
    capsize=5,
    capthick=0.8,

    linestyle="none",
    zorder=3
)
#SH0ES REFERENCE LINE
SH0ES = 73.04

ax.axhline(
    SH0ES,
    color="#c83b32",
    linestyle=":",
    linewidth=1.5,
    zorder=1
)


ax.text(
    6.02,
    SH0ES + 0.05,
    r"SH0ES (R22)",
    color="#c83b32",
    fontsize=12,
    va="bottom",
    ha="left"
)

ax.text(
    -0.30,
    71.35,
    r"$Planck:\ 67.4\pm0.5$",
    color="#234b7d",
    fontsize=12,
    fontstyle="italic",
    ha="left",
    va="center"
)

ax.text(
    -0.30,
    70.75,
    r"$(below\ plot\ range)$",
    color="#234b7d",
    fontsize=12,
    fontstyle="italic",
    ha="left",
    va="center"
)

labels = [
    r"$R_W=0.386$" + "\n" + r"$\mathrm{fixed}$",

    r"$\mathrm{Global}$" + "\n" + r"$R_W$",

    r"$\mathrm{Indiv.}\ b_{W,i}$" + "\n" + r"$(\mathrm{fixed}\ R_W)$",

    r"$R_E=0.386$" + "\n" + r"$\mathrm{fixed}$",

    r"$\mathrm{Global}$" + "\n" + r"$R_E$",

    r"$\mathrm{Indiv.}\ R_E$" + "\n" + r"$[0,1]$",

    r"$\mathrm{Indiv.}\ R_E$" + "\n" + r"$[0.15,0.8]$"
]

ax.set_xticks(x)
ax.set_xticklabels(
    labels,
    fontsize=11
)

ax.set_xlim(-0.55, 6.55)
ax.set_ylim(70.5, 79.5)

ax.set_yticks(np.arange(71, 80, 1))

ax.set_ylabel(
    r"$H_0\;(\mathrm{km\,s^{-1}\,Mpc^{-1}})$",
    fontsize=13,
    labelpad=8
)

ax.tick_params(
    axis="y",
    labelsize=10,
    direction="out"
)

ax.grid(
    True,
    which="major",
    axis="both",
    color="#d9d9d9",
    linewidth=0.6,
    alpha=0.65
)

ax.set_axisbelow(True)




ax.set_title(
    r"$H_0$ under Different Cepheid Calibration Variants "
    r"(Mörtsell et al. 2022 [$\mathrm{[2]}$])",
    fontsize=15,
    pad=12
)

for spine in ax.spines.values():
    spine.set_linewidth(0.8)
    spine.set_color("#444444")



ax.tick_params(
    axis="x",
    which="major",
    length=0,
    pad=8
)

ax.tick_params(
    axis="y",
    which="major",
    length=3,
    width=0.7
)
plt.subplots_adjust(
    left=0.17,
    right=0.875,
    top=0.82,
    bottom=0.23
)




plt.savefig(
    "mortsell_2022_fig2_reproduced.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
